# 01. Intraday Market Data Exploration

This notebook inspects canonical 5-minute OHLCV market bars, checks data geometry, and validates timestamp distributions.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from alpha_volatility.data.storage import ParquetStorage
from alpha_volatility.data.validation import validate_bars

storage = ParquetStorage('data/synthetic')
df = storage.load('synthetic_bars.parquet')
print(f'Total bars loaded: {len(df):,}')
print(f'Unique assets: {df["asset_id"].nunique()}')
df.head()
```

```python
report = validate_bars(df)
print(f'Dataset validation valid: {report.is_valid}')
print(f'Errors: {len(report.errors)}, Warnings: {len(report.warnings)}')
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
for asset in df['asset_id'].unique()[:3]:
    sub = df[df['asset_id'] == asset].sort_values('timestamp')
    ax.plot(sub['timestamp'], sub['close'], label=asset)
ax.set_title('Sample Intraday Close Price Trajectories')
ax.legend()
plt.show()
```
